```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
```

```python
class BatsmanState(TypedDict):
    runs: int
    balls: int 
    fours: int
    sixes: int 
    
    sr: float
    bpb: float
    boundry_pct: float
    summary : str
```

```python
def calculate_sr(state:BatsmanState):
    runs= state['runs']
    balls = state['balls']

    sr = round(runs/balls, 2)


    return {'sr':sr}
```

```python
def calculate_bpb(state: BatsmanState):
    fours = state['fours']
    sixes= state['sixes']
    balls = state['balls']
    bpb = round((fours + sixes)/ balls,2)

    

    return {'bpb': bpb}
```

```python
def calculate_boundry_pct(state : BatsmanState):
        fours = state['fours']
        sixes= state['sixes']
        runs = state['runs']

        runs_in_boundry = fours*4 + sixes* 6 

        boundry_pct= round(runs_in_boundry/runs , 2) 

        
        return {'boundry_pct': boundry_pct}
```

```python
def summary(state:BatsmanState)-> BatsmanState:
    summary = f"""
Strike Rate - {state['sr']} \n
Balls per boundry - {state['bpb']} \n
Boundry Percentage - {state['boundry_pct']} \n 
"""

    state['summary']= summary

    return state 
```

```python
graph = StateGraph(BatsmanState)

graph.add_node('calculate_sr', calculate_sr)
graph.add_node('calculate_bpb', calculate_bpb)
graph.add_node('calculate_boundry_pct', calculate_boundry_pct)
graph.add_node('summary', summary )


graph.add_edge(START, 'calculate_sr')
graph.add_edge(START, 'calculate_bpb')
graph.add_edge(START, 'calculate_boundry_pct')

graph.add_edge(['calculate_boundry_pct', 'calculate_bpb', 'calculate_sr'], 'summary')
graph.add_edge('summary', END)

workflow= graph.compile()
```

```python
initial_state = {'runs': 48, 'balls':28, 'fours': 3, 'sixes':4}
final_state = workflow.invoke(initial_state)

print(final_state['summary'])
```

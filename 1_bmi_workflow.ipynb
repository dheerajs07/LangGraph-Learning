```python
from langgraph.graph import StateGraph, START , END
from typing import TypedDict
```

```python
# Define State
class BMIState(TypedDict):
    weight_kg: float
    height_m : float
    bmi : float
    category: str
```

```python
def calculate_bmi(state : BMIState) -> BMIState:
    weight = state['weight_kg']
    height = state['height_m']
    bmi = weight/(height** 2 )

    state['bmi']= round(bmi, 2)

    return state

def check_category(state: BMIState)-> BMIState:
    bmi = state['bmi']
    if(bmi<18.5):
        state['category']= "Underweight"
    elif (18.8<= bmi <25) :
        state['category']= "Normal"
    elif(25<=bmi<30):
        state['category']= "Overweight"
    else:
        state['category']="Obese"
    return state
```

```python
# define your graph
graph = StateGraph(BMIState)

# add nodes to the graph 
graph.add_node('calculate_bmi', calculate_bmi)
graph.add_node('check_category', check_category)
# add edges to your graph 
graph.add_edge(START, 'calculate_bmi')
graph.add_edge('calculate_bmi', 'check_category')
graph.add_edge('check_category', END)

# compile the graph 
workflow= graph.compile()

# execute the graph 
workflow.invoke({'weight_kg': 85 , 'height_m': 1.8})
```
